```python
data=["1 Pen 20","2 Pencil 30","3 Pen 5","4 Pencil 50"]
```

```python
#Fetch the total Qty of Pen

penItem=filter(lambda i: i.split()[1]=="Pen",data)
totalQty = sum(map(lambda i: int(i.split()[-1]), penItem))
print(totalQty)
```

```python
try:
    
    try:
        num1=int(input("Enter the number 1"))
    except ValueError:
        num1=int(input("Enter the number 1 again.."))
        
    try:
        num2=int(input("Enter the number 1"))
    except ValueError:
        num2=int(input("Enter the number again "))
    
    res=num1/num2
    

except ZeroDivisionError:
    print("You divided by zero, which is wrong")
    num2=int(input("Enter the number 2 again except zero"))
    res=num1/num2

except:
    print("This is a new error")
    
finally:
    print("This will always execute")
    
    

    
```
