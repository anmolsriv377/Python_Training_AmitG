```python
# Take no. as input from user. if no. divisible by 5 print Hi, if divisible by 7 print bye
# if divisible by 5 and 7 both print HiBye, else print NA

num = int(input("Enter no. : "))
if(num%5==0 and num%7==0):
    print("HiBye")
elif(num%5==0):
    print("Hi")
elif(num%7==0):
    print("Bye")
else:
    print("NA")
```
