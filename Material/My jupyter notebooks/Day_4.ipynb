```python
# Add a new column in dataframe
bank["Interest"]=bank["Balance"]*0.1
```

```python
bank
```

```python
bank.to_csv("bank_interest.csv")
```

```python
# Drop a column for row or column 
# axis=0 if row
# axis=1 if column
bank.drop("Interest",axis=1, inplace=True)
```

```python
bank
```

```python
# Remove the blank rows
bank.dropna(inplace=True)
```

```python
bank=bank.loc[bank.Region!="England"]
```

```python
# add the column of interest -- balance>100000 --10%
# Balance 50000-1L --8%
# <50000 ---6%
def interest(b):
    if b>100000:
        return 0.1*b
    elif b>=50000 and b<=100000:
        return b*0.08
    else:
        return b*0.06
```

```python
bank["Interest"] = bank["Balance"].apply(interest)
```

```python
bank
```

```python
# introduce some null values
bank.loc[bank.Age>60,"Age"]=np.nan
```

```python
bank
```

```python
bank.count()
```

```python
# % of missing values in each column
round((1-bank.count()/bank.shape[0])*100,2)
```

```python
bank.loc[bank.Age.isna(),:]
bank.loc[bank.Age.notna(),:]
```

```python
bank.Age.mean()
```

```python
# Age is null where put avg age
# in the numpy lib we have where
bank["Age"] = np.where(bank.Age.isna(), bank.Age.mean(), bank.Age)
```
