```python
import unittest
```

```python
# Create a function that calculates the area of circle
def Circle(r):
    if type(r) not in [int,float]:
        raise TypeError("The radius ust be a number")
    if r<0:
        raise ValueError("The radius cannot be negative")
    return 3.14*r*r
```

```python
class TestMath(unittest.TestCase):
    #Test areas when radius>0
    def test_area(self):
        self.assertAlmostEqual(Circle(1),3.14)
        self.assertAlmostEqual(Circle(0),0)
        self.assertAlmostEqual(Circle(2.1), 3.14*2.1*2.1)

    # Negative values
    def test_values(self):
        self.assertRaises(ValueError,Circle,-2)

    def test_types(self):
        self.assertRaises(TypeError,Circle, 2+3j)
        self.assertRaises(TypeError,Circle, True)
        self.assertRaises(TypeError,Circle, "radius")
```

```python
unittest.main(argv=[''], verbosity=2, exit=False)
```
