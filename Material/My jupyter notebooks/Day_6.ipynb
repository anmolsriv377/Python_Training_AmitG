```python
# Create an object Circle.. the object should have 2 attributes.. radius and area
# It should have 2 methods .. circumference .. change radius
```

```python
class Circle:
    def __init__(self, radius):
        self.radius=radius
        self.area=self.radius*self.radius*3.14

    def changeRadius(self, newRadius):
        self.radius=newRadius
        self.area=newRadius*newRadius*3.14

    def circumference(self):
        return 2*3.14*self.radius
```

```python
x=Circle(10)
```

```python
x.radius
```

```python
x.area
```

```python
x.changeRadius(15)
```

```python
x.radius
```

```python
x.area
```

```python
# Create an object Accounts
# The attributes in the object are name and balance
# 2 Methods - withdrawal and deposit...
    # if deposit some amount my balance should update and print a message .. deposit successful
    # if I withdraw then if the withdrawal amount is more I should get a message insufficient funds
    # other update the balance after deposit and print (deposit successful)

class Accounts:
    def __init__(self, name, balance):
        self.name = name
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print("deposit successful")

    def withdrawal(self, amount):
        if amount > self.balance:
            print("Insufficient funds")
        else:
            self.balance -= amount
            print("Withdrawl Successfully")
```

```python
x = Accounts("Sam", 1000)
```

```python
x.deposit(100)
```

```python
x.balance
```

```python
x.withdrawal(500)
```

```python
x.balance
```

```python
# When a pattern is defined
# identifiers
# \d --> looking for a digit
# \w --> looking fror alphanumeric
# \s --> looking for white space
# \D --> looking for A non digit
# \W --> looking for a non alphanumeric
# \S --> looking for a non whitespace

text = "The agent's phone number is 408-555-1234. Call Soon"
```

```python
re.search("\d\d\d-\d\d\d-\d\d\d\d", text)
```

```python
# Quantifiers
# + --> occurs once or more times
# {3} --> occurs exactly three times
# {2,4} --> occurs two to four times
# {3, } --> occurs three or more times
# * --> occurs zero or more times
# ? --> occurs once or none times
```

```python
text = "The agent's phone number is 408-555-1234. Call Soon"
re.search("\d\{3}-\d{3}-\d{4}", text)
```
