```python
# Decorator Functions

# A function... And you want to use that function with other functions...
# decorators

import time
def brew_tea():
    start_time=time.time()
    print("Brewing tea....")
    time.sleep(1)
    print("Tea is ready")
    end_time=time.time()
    print("Task Time", end_time-start_time)
```

```python
# so there is redundancy of code
def make_matcha():
    print("Making mactha....")
    time.sleep(2)
    print("Matcha is ready")
```

```python
# Decorator time function
def timer_dec(base_fn):
    def enhanced_fn():
        start_time=time.time()
        base_fn()
        end_time=time.time()
        print("Task Time", end_time-start_time)
    return enhanced_fn()
```

```python
timer_dec(make_matcha)
```

```python
x=timer_dec(make_matcha)
```

```python
x()
```

```python
@timer_dec
def brew_tea():
    print("Brewing tea....")
    time.sleep(1)
    print("Tea is ready")
```
