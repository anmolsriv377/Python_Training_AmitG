```python
import numpy as np
import pandas as pd
```

```python
bank = pd.read_csv("P1-UK-Bank-Customers (1) (8).csv")
```

```python
# Aggregations
bank.Balance     # SQL way
bank["Balance"]
# Both are same in dataframes  
bank["Balance"].sum()
```

```python
bank["Balance"].mean()
```

```python
# Groupby - for a summary of data
# Region wise sum of balance
bank.groupby("Region")["Balance"].sum()
```

```python
# To convert to dataframe
# It resets the index and puts default indexing and converts Series to Dataframe
bank.groupby("Region")["Balance"].sum().reset_index()
```

```python
# Value_counts
bank["Region"].value_counts()
```

```python
# Gender wise average age
bank.groupby("Gender")["Age"].mean()
```

```python
# To convert to dataframe
bank.groupby("Gender")["Age"].mean().reset_index()
```

```python
# To group data by 2 levels on basis of balance
bank.groupby(["Region","Gender"])["Balance"].sum()
```

```python
bank.groupby(["Region","Gender"])["Balance"].sum().reset_index()
```

```python
# Multiple Aggregation
# Region wise Gender wise
bank.groupby("Region")[["Balance","Age"]].sum()
# region wise sum of balance , sum of age
# But here we generally don't calculate age sum we calculate avg. age
```

```python

type(bank.groupby(["Region","Gender"])["Balance"].sum())
```

```python

type(bank.groupby(["Region","Gender"])["Balance"].sum().reset_index())
```

```python
# Here we get sum and mean of balance and age both using agg function
bank.groupby("Region")[["Balance","Age"]].agg(["sum","mean"])
```

```python
# Now we get sum of balance and mean of age using dictionary defining using key value pairs
bank.groupby("Region")[["Balance","Age"]].agg({"Balance":"sum","Age":"mean"})
```

```python
# WAP to fetch the avg. balance by job classification for the region where 
# we have highest avg balance
print(bank.columns)
```

```python
# Region with highest avg balance
x = bank.groupby("Region")["Balance"].mean().idxmax()
# or 
r=bank.groupby("Region")["Balance"].mean().sort_values(ascending=False).reset_index().iloc[0,0]
```

```python
bank.loc[bank.Region==r, :].groupby(["Region","Job Classification"])["Balance"].mean()
```

```python
# Pivot
summ = bank.groupby(["Region","Gender"])["Balance"].sum().reset_index()
```

```python
summ.pivot(index="Region",columns="Gender")
```

```python
bank.pivot_table(index="Region",columns="Gender",values="Balance",aggfunc="sum")
```

```python
# Load the order sheets from the excel 
orders = pd.read_excel("Sample - Superstore (1) (1) (1).xlsx", sheet_name="Orders")
```

```python
# Tasks
# 1. Fetch the unique category
# 2. Fetch the unique category and sub category
# 3. Fetch the top 5 selling products by sales
# 4. Create a crosstab to show sales by categroy and Region
# 5. Fetch the custmer Name , Sales for category "Furniture" and Region "East"
```

```python
orders.columns
```

```python
# 1. Fetch the unique category
# orders["Category"].unique()
orders.Category.unique()
```

```python
# 2. Fetch the unique category and sub category
# Unique() only works on one category and not two so we can use drop_duplicates()
orders[["Category","Sub-Category"]].drop_duplicates()
# loc gives data in row and columns 
orders.loc[:, ["Category", "Sub-Category"]].drop_duplicates()
```

```python
# 3. Fetch the top 5 selling products by sales
orders.groupby("Product Name")["Sales"].sum().sort_values(ascending=False).reset_index().head()
```

```python
# 4. Create a crosstab to show sales by categroy and Region

orders.pivot_table(index="Category", columns="Region", values="Sales", aggfunc="sum")
```

```python
# 5. Fetch the custmer Name , Sales for category "Furniture" and Region "East"
orders.loc[(orders.Region=="East")&(orders.Category=="Furniture"), ["Customer Name","Sales"]]
```
