```python
def jamesbond(L):
    comp=[0,0,7,'X']
    for i in L:
        if i==comp[0]:
            comp.pop(0)
    if len(comp)==1:
        return "James Bond"
    else:
        return False
        
```

```python
jamesbond([1,2,7,2,0,0])
```

```python
def start_game():
    print("Welcome to the game")
    name=input("Please enter your name")
    print("Hello", name, "Your mission is to find a hidden treasure")
    print("Where do you want to go")
    print("1. Forest")
    print("2. Cave")

    choice=int(input("Enter your choice as 1 or 2"))
    if choice ==1:
        forest()
    elif choice==2:
        cave()
    else:
        print("Invalid Choice")
        start_game()
        
        

def forest():
    print("Welcome to the forest")
    print("You are in a forest and you see an animal")
    print("1. Climb a tree")
    print("2. Run")
    choice=int(input("Enter your choice as 1 or 2"))
    
    if choice==1:
        print("You won the game")
    else:
        print("You Lost")

def cave():
    print("Welcome to the cave")
    print("You are in a cave and its dark")
    print("1. Light a torch")
    print("2. Run")
    choice=int(input("Enter your choice as 1 or 2"))
    
    if choice==1:
        print("You won the game")
    else:
        print("You Lost")
```
